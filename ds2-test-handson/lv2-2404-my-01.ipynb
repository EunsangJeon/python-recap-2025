```python
df_1['REGION'].value_counts().reset_index().sort_values(by='REGION').reset_index(drop=True)
```

```python
rt = pd.DataFrame()
```

```python
rt['region'] = df_1.groupby('REGION')['TENURE'].mean().reset_index()['REGION']
rt['mean'] = df_1.groupby('REGION')['TENURE'].mean().reset_index()['TENURE']
```

```python
rt['std'] = df_1.groupby('REGION')['TENURE'].std().reset_index()['TENURE']
```

```python
rt['cnt'] = df_1['REGION'].value_counts().reset_index().sort_values(by='REGION').reset_index(drop=True)['count']
```

```python
rt
```

```python
df_1['REGION'].value_counts()
```

# 1-2

```python
t_range = {'290': 1.968178, '295': 1.968038, '312': 1.967596}
```

```python
min_t = []

min_t.append(rt.loc[0, 'mean'] - t_range[str(rt.loc[0, 'cnt'] - 1)] * rt.loc[0, 'std'] / (rt.loc[0, 'cnt'] ** 0.5))
min_t.append(rt.loc[1, 'mean'] - t_range[str(rt.loc[1, 'cnt'] - 1)] * rt.loc[1, 'std'] / (rt.loc[1, 'cnt'] ** 0.5))
min_t.append(rt.loc[2, 'mean'] - t_range[str(rt.loc[2, 'cnt'] - 1)] * rt.loc[2, 'std'] / (rt.loc[2, 'cnt'] ** 0.5))
```

```python
min_t
```

```python
max_t = []

max_t.append(rt.loc[0, 'mean'] + t_range[str(rt.loc[0, 'cnt'] - 1)] * rt.loc[0, 'std'] / (rt.loc[0, 'cnt'] ** 0.5))
max_t.append(rt.loc[1, 'mean'] + t_range[str(rt.loc[1, 'cnt'] - 1)] * rt.loc[1, 'std'] / (rt.loc[1, 'cnt'] ** 0.5))
max_t.append(rt.loc[2, 'mean'] + t_range[str(rt.loc[2, 'cnt'] - 1)] * rt.loc[2, 'std'] / (rt.loc[2, 'cnt'] ** 0.5))
```

```python
max_t
```

# 1-3

```python
A = max(max_t)
B = min(min_t)
A + B
```

# Q1 Answer

```python
print(round(A + B, 2))
```

# Q2

```python
df_2 = CUSTOMER.copy()
```

# 2-1

```python
df_2.columns
```

```python
targets = ['TENURE', 'AGE', 'INCOME', 'EMPLOY']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
for target in targets:
    df_2[target + '_S'] = StandardScaler().fit_transform(X=df_2.loc[:, [target]])
```

```python
df_2
```

# 2-2

```python
from sklearn.cluster import KMeans
```

```python
targets = ['TENURE_S', 'AGE_S', 'INCOME_S', 'EMPLOY_S']
```

```python
model = KMeans(n_clusters=4, random_state=1234).fit(X=df_2.loc[:, targets])
```

# 2-3

```python
model.cluster_centers_
```

```python
len(model.cluster_centers_)
```

# 2-4

```python
centres = model.cluster_centers_
```

```python
df_2['LABEL'] = model.predict(X = df_2.loc[:, targets])
```

```python
dfs = []

for i in range(4):
    df = df_2[df_2['LABEL'] == i].copy()
    df['DIS'] = ((df['TENURE_S'] - centres[i][0]) ** 2 + (df['AGE_S'] - centres[i][1]) ** 2 + (df['INCOME_S'] - centres[i][2]) ** 2 + (df['EMPLOY_S'] - centres[i][3]) ** 2) ** 0.5
    dfs.append(df)
```

```python
dfs
```

# 2-5

```python
stds = []

for df in dfs:
    stds.append(df['DIS'].std())
```

```python
stds
```

```python
min(stds)
```

# Q2 Answer

```python
print(round(min(stds), 3))
```

# Q3

```python
df_3 = CUSTOMER.copy()
```

# 3-1

```python
DF23 = df_3[(df_3['CUSTCAT'] == 'Class A') | (df_3['CUSTCAT'] == 'Class D')].copy()
```

# 3-2

```python
ages = [50, 40, 30]
```

```python
DF23['AGE_GROUP'] = DF23['CUS_ID'].map(lambda x: 'AGE_50')
```

```python
for age in ages:
    DF23['AGE_GROUP'] = np.where((DF23['AGE'] < age), 'AGE_' + str(age - 10),  DF23['AGE_GROUP'])
```

```python
DF23['AGE_GROUP'].value_counts()
```

# 3-3

```python
DF23.head()
```

```python
DF23.columns
```

```python
targets = ['REGION', 'ED', 'AGE_GROUP']
```

```python
DF23 = pd.get_dummies(DF23, drop_first=True, columns=targets, dtype='int')
```

```python
DF23.head()
```

# 3-4

```python
TRAIN23 = DF23[DF23['CUS_ID'] % 5 != 0]
TEST23 = DF23[DF23['CUS_ID'] % 5 == 0]
```

```python
TRAIN23.shape
```

```python
TEST23.shape
```

# 3-5

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
TRAIN23.columns
```

```python
targets = ['TENURE', 'INCOME', 'MARITAL', 'EMPLOY', 'RETIRE', 'GENDER', 'REGION_2', 'REGION_3', 'ED_2', 'ED_3', 'ED_4', 'ED_5', 'AGE_GROUP_AGE_30', 'AGE_GROUP_AGE_40', 'AGE_GROUP_AGE_50']
```

```python
len(targets)
```

```python
model = DecisionTreeClassifier(max_depth=3, random_state=1234).fit(y=TRAIN23['CUSTCAT'], X=TRAIN23.loc[:, targets])
```

# 3-6

```python
from sklearn.metrics import precision_score
```

```python
pred = model.predict(X=TEST23.loc[:, targets])
```

```python
precision_score(y_true=TEST23['CUSTCAT'], y_pred=pred, pos_label='Class A')
```

# Q3 Answer

```python
print(round(precision_score(y_true=TEST23['CUSTCAT'], y_pred=pred, pos_label='Class A'), 3))
```
