# Q1

```python
df_1 = SALES.copy()
```

# 1-1

```python
df_1.columns
```

```python
targets = ['MD1', 'MD2', 'MD3', 'TEMP', 'FUEL', 'CPI', 'UNEMP']
```

```python
len(targets)
```

```python
from scipy.stats import pearsonr
```

```python
ps = []

for target in targets:
    p = pearsonr(x=df_1[target], y=df_1['SALES'])
    ps.append(abs(p[0]))
```

```python
ps
```

# 1-2

```python
max_ps = max(ps)
```

```python
max_ps
```

# Q1 Answer

```python
print(round(max_ps, 3))
```

```python
targets = ['SALES', 'MD1', 'MD2', 'MD3', 'TEMP', 'FUEL', 'CPI', 'UNEMP']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
for target in targets:
    DF22[target + '_S'] = StandardScaler().fit_transform(X=DF22.loc[:, [target]])
```

# 2-3

```python
DF22.columns
```

```python
targets = ['SALES_S', 'MD1_S', 'MD2_S', 'MD3_S', 'TEMP_S', 'FUEL_S', 'CPI_S', 'UNEMP_S']
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=4, random_state=1234, init='random').fit(X=DF22.loc[:, targets])
```

```python
DF22['LABEL'] = model.predict(X=DF22.loc[:, targets])
```
