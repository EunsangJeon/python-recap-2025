```python
import pandas as pd
```

```python
EXPT = pd.read_csv('lv2-2410.csv')
```

```python
EXPT.shape
```

```python
EXPT.head()
```

# Q1

```python
df_1 = EXPT.copy()
```

# 1-1

```python
df_1.columns
```

```python
targets = ['MOTOR_SPEED', 'F1', 'F2', 'F3', 'F4', 'F5', 'F6', 'F7', 'F8', 'F9', 'F10', 'F11']
```

```python
len(targets)
```

```python
cvs = []

for target in targets:
    s = df_1[target].std()
    m = df_1[target].mean()
    cv = s / m
    cvs.append(cv)
```

```python
cvs
```

# 1-2

```python
abs_cvs = []

for cv in cvs:
    abs_cvs.append(abs(cv))
```

```python
abs_cvs
```

```python
max(abs_cvs)
```

# Q1 Answer

```python
print(round(max(abs_cvs), 2))
```

# Q2

```python
df_2 = EXPT.copy()
```

# 2-1

```python
df_2['HIGH_SPEED'] = (df_2['MOTOR_SPEED'] > 1) + 0
```

```python
df_2.head()
```

# 2-2

```python
TRAIN25 = df_2[df_2['PID'] % 5 != 0].copy()
TEST25 = df_2[df_2['PID'] % 5 == 0].copy()
```

```python
TRAIN25.shape
```

```python
TEST25.shape
```

# 2-3

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
df_2.columns
```

```python
targets = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6', 'F7', 'F8', 'F9', 'F10', 'F11']
```

```python
model = DecisionTreeClassifier(max_depth=3, random_state=1234).fit(y=TRAIN25['HIGH_SPEED'], X=TRAIN25.loc[:, targets])
```

# 2-4

```python
pred = model.predict(X=TEST25.loc[:, targets])
```

```python
from sklearn.metrics import recall_score
```

```python
rec = recall_score(y_true=TEST25['HIGH_SPEED'], y_pred=pred)
```

```python
rec
```

# Q2 Answer

```python
print(round(rec, 3))
```

# Q3

```python
df_3 = EXPT.copy()
```

# 3-1

```python
targets = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6', 'F7', 'F8', 'F9', 'F10', 'F11']
```

```python
df_3['TMP'] = df_3['PID'].map(lambda x: 0)
```

```python
df_3.head()
```

```python
for target in targets:
    df_3['TMP'] += (df_3[target] > 1.5) + 0
```

```python
df_3['TMP'].value_counts()
```

```python
DF26 = df_3[df_3['TMP'] == 0]
```

```python
DF26 = DF26.drop(columns=['TMP'])
```

```python
DF26.shape
```

```python
DF26.head()
```

# 3-2

```python
TRAIN26 = DF26[DF26['PID'] % 6 != 0]
TEST26 = DF26[DF26['PID'] % 6 == 0]
```

```python
TRAIN26.shape
```

```python
TEST26.shape
```

# 3-3

```python
from sklearn.linear_model import LinearRegression
```

```python
targets = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6', 'F7', 'F8', 'F9', 'F10', 'F11']
```

```python
model = LinearRegression().fit(y=TRAIN26['MOTOR_SPEED'], X=TRAIN26.loc[:, targets])
```

# 3-4

```python
pred = model.predict(X=TEST26.loc[:, targets])
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae = mean_absolute_error(y_true=TEST26['MOTOR_SPEED'], y_pred=pred)
```

```python
mae
```

# Q3 Answer

```python
print(round(mae, 3))
```
