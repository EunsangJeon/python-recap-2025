```python
import pandas as pd
import numpy as np
```

```python
FAC = pd.read_csv('lv2-2405.csv')
```

# Q1

```python
df_1 = FAC.copy()
```

# 1-1

```python
FAC0 = df_1[df_1['FAIL_YN'] == 0].copy()
```

```python
FAC0.shape
```

```python
FAC1 = df_1[df_1['FAIL_YN'] == 1].copy()
```

```python
FAC1.shape
```

# 1-2

```python
cvs_0 = []
cvs_1 = []
```

```python
FAC0.columns
```

```python
targets = ['TRQ_PWR', 'DUR', 'NS', 'TEMP_DIFF']
```

```python
len(targets)
```

```python
for target in targets:
    s_0 = FAC0[target].std()
    x_0 = FAC0[target].mean()
    cv_0 = s_0 / x_0

    s_1 = FAC1[target].std()
    x_1 = FAC1[target].mean()
    cv_1 = s_1 / x_1

    cvs_0.append(cv_0)
    cvs_1.append(cv_1)
```

# 1-3

```python
gaps = []

for i in range(4):
    gaps.append(abs(cvs_0[i] - cvs_1[i]))
```

```python
gaps
```

```python
max(gaps)
```

# Q1 Answer

```python
print(round(max(gaps), 3))
```

# 3-2

```python
TRAIN6 = df_3[df_3['PRD_ID'] % 3 != 0]
TEST6 = df_3[df_3['PRD_ID'] % 3 == 0]
```

```python
TRAIN6.shape
```

```python
TEST6.shape
```

# 3-3

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
df_3.columns
```

```python
targets = ['ROT_SPD', 'TRQ_PWR', 'DUR', 'NS', 'TEMP_DIFF']
```

```python
model = DecisionTreeClassifier(random_state=3, min_samples_split=3).fit(y=TRAIN6['FAIL_YN'], X=TRAIN6.loc[:, targets])
```

```python
models = []
ks = [3, 5, 7]

for k in ks:
    model = DecisionTreeClassifier(random_state=3, min_samples_split=3, max_depth=k).fit(y=TRAIN6['FAIL_YN'], X=TRAIN6.loc[:, targets])
    models.append(model)
```

# 3-4

```python
recs = []
```

```python
from sklearn.metrics import recall_score
```

```python
for model in models:
    pred = model.predict(X=TRAIN6.loc[:, targets])
    recs.append(recall_score(y_true=TRAIN6['FAIL_YN'], y_pred=pred))
```

```python
recs
```
