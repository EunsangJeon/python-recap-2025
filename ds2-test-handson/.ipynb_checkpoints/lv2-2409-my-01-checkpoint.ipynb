# Q1

```python
df_1 = INSPECTION.copy()
```

```python
diffs = []
```

```python
df_1.columns
```

# 1-1

```python
targets = ['P01', 'P02', 'P03', 'P04', 'P05', 'P06', 'P07', 'P08', 'P09', 'P10', 'P11']
```

```python
df_1_0 = df_1[df_1['DEFECT'] == 0].copy()
df_1_1 = df_1[df_1['DEFECT'] == 1].copy()
```

```python
df_1_0.shape
```

```python
df_1_1.shape
```

```python
for target in targets:
    Q01 = df_1_0[target].quantile(q=0.25)
    Q02 = df_1_0[target].quantile(q=0.50)
    Q03 = df_1_0[target].quantile(q=0.75)
    Q11 = df_1_1[target].quantile(q=0.25)
    Q12 = df_1_1[target].quantile(q=0.50)
    Q13 = df_1_1[target].quantile(q=0.75)

    diff = abs(Q01-Q11) + abs(Q02-Q12) + abs(Q03-Q13)

    diffs.append(diff)
```

```python
diffs
```

```python
len(diffs)
```

# 1-2

```python
max(diffs)
```

# Q1 Answer

```python
print(round(max(diffs)))
```

# Q3

```python
df_3 = INSPECTION.copy()
```

# 3-1

```python
targets = ['P01', 'P02', 'P03', 'P04', 'P05', 'P06', 'P07', 'P08', 'P09', 'P10', 'P11']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
for target in targets:
    df_3[target + '_S'] = StandardScaler().fit_transform(X=df_3.loc[:, [target]])
```

```python
df_3.head()
```

# 3-2

```python
TRAIN23 = df_3[df_3['PROD_ID'] % 5 != 0].copy()
TEST23 = df_3[df_3['PROD_ID'] % 5 == 0].copy()
```

```python
TRAIN23.shape
```

```python
TEST23.shape
```

# 3-3

```python
ks = [3, 5, 7]
models = []
```

```python
tmp = []

for target in targets:
    tmp.append(target + '_S')
```

```python
targets = tmp
```

```python
targets
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
for k in ks:
    model = KNeighborsClassifier(n_neighbors=k, metric='euclidean').fit(y=TRAIN23['DEFECT'], X=TRAIN23.loc[:, targets])
    models.append(model)
```

# 3-4

```python
accs = []
```

```python
from sklearn.metrics import accuracy_score
```

```python
for model in models:
    pred = model.predict(X=TEST23.loc[:, targets])
    accs.append(accuracy_score(y_true=TEST23['DEFECT'], y_pred=pred))
```

```python
accs
```

# 3-5

```python
ans = pd.DataFrame(accs, columns=['acc'])
```

```python
ans['acc'].mean()
```

```python
ans
```

# Q3 Answer

```python
print(round(ans['acc'].mean(), 3))
```
