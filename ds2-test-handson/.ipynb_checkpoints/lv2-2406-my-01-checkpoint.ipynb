# Q1

```python
df_1 = REF.copy()
```

```python
df_1.columns
```

# 1-1

```python
from math import log2
```

```python
targets = ['PART02', 'PART03', 'PART04', 'PART05', 'PART06']
```

```python
ref_count = df_1.shape[0]
```

```python
df_1[df_1['PART02'] == 'A']['PRD_ID'].count()
```

```python
ents = []

for target in targets:
    c_a = df_1[df_1[target] == 'A']['PRD_ID'].count()
    c_b = df_1[df_1[target] == 'B']['PRD_ID'].count()
    c_a_g = df_1[(df_1[target] == 'A') & (df_1['CLASS'] == 'G')]['PRD_ID'].count()
    c_a_b = df_1[(df_1[target] == 'A') & (df_1['CLASS'] == 'B')]['PRD_ID'].count()
    c_b_g = df_1[(df_1[target] == 'B') & (df_1['CLASS'] == 'G')]['PRD_ID'].count()
    c_b_b = df_1[(df_1[target] == 'B') & (df_1['CLASS'] == 'B')]['PRD_ID'].count()
    p_a_g = c_a_g / c_a
    p_a_b = c_a_b / c_a
    p_b_g = c_b_g / c_b
    p_b_b = c_b_b / c_b
    e_a = (p_a_g * log2(p_a_g) + p_a_b * log2(p_a_b)) * -1
    e_b = (p_b_g * log2(p_b_g) + p_b_b * log2(p_b_b)) * -1
    entropy = (c_a / ref_count * e_a) + (c_b / ref_count * e_b)
    ents.append(entropy)
```

```python
ents
```

# 1-2

```python
max(ents) + min(ents)
```

# Q1 Answer

```python
print(round(max(ents) + min(ents), 2))
```

```python
df_2 = REF.copy()
```

```python
df_2.columns
```

# 2-1

```python
targets = ['PART00', 'PART01', 'PART07', 'PART08', 'PART09']
```

```python
df_2 = pd.get_dummies(df_2, drop_first=True, columns=targets, dtype='int')
```

```python
df_2.head()
```

```python
df_2.columns
```

```python
df_2.columns[-23:]
```

# 2-2

```python
targets = df_2.columns[-23:]
```

```python
targets
```

```python
g_effs = []

for target in targets:
    prob_g0 = df_2[(df_2['CLASS'] == 'G') & (df_2[target] == 0)][target].count() / df_2[df_2[target] == 0][target].count()
    prob_g1 = df_2[(df_2['CLASS'] == 'G') & (df_2[target] == 1)][target].count() / df_2[df_2[target] == 1][target].count()
    g_eff = max(prob_g1/(prob_g0 + 0.01), prob_g0/(prob_g1 + 0.01))
    g_effs.append(g_eff)
```

```python
len(g_effs)
```

```python
g_effs[:5]
```

# 2-3

```python
g_effs.sort(reverse=True)
```

```python
len(g_effs[:10])
```

```python
g_effs[:10]
```

```python
import numpy as np
```

```python
np.mean(g_effs[:10])
```

# Q2 Answer

```python
print(round(np.mean(g_effs[:10]), 3))
```

# Q3

```python
df_3 = REF.copy()
```

```python
df_3.columns
```

# 3-1

```python
targets = ['PART10', 'PART11', 'PART12', 'PART13']
```

```python
df_3 = pd.get_dummies(data=df_3, columns=targets, drop_first=True, dtype='int')
```

```python
df_3.head()
```

```python
df_3.columns[-14:]
```

# 3-2

```python
TRAIN23 = df_3[df_3['PRD_ID'] % 5 != 0]
TEST23 = df_3[df_3['PRD_ID'] % 5 == 0]
```

```python
TRAIN23.shape
```

```python
TEST23.shape
```

# 3-3

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier(random_state=1234, max_depth=9).fit(y=TRAIN23['CLASS'], X=TRAIN23.iloc[:, -14:])
```

# 3-4

```python
model.feature_importances_
```

```python
fi = pd.DataFrame(data=df_3.columns[-14:], columns=['feature'])
```

```python
fi
```

```python
fi['fi'] = model.feature_importances_
```

```python
fi
```

```python
VAR23 = fi.sort_values(by=['fi'], ascending=False)
```

```python
VAR23
```

# 3-5

```python
models = []
```

```python
target_l = [5, 10, 14]
```

```python
for l in target_l:
    model  = DecisionTreeClassifier(random_state=1234, max_depth=9).fit(y=TRAIN23['CLASS'], X=TRAIN23.loc[:, VAR23['feature'][:l]])
    models.append(model)
```

```python
len(models)
```

# 3-6

```python
accs = []
```

```python
from sklearn.metrics import accuracy_score
```

```python
index = 0
for model in models:
    l = target_l[index]
    pred = model.predict(X=TEST23.loc[:, VAR23['feature'][:l]])
    acc = accuracy_score(y_true=TEST23['CLASS'], y_pred=pred)
    accs.append(acc)
    index += 1
```

```python
accs
```

# Q3 Answer

```python
print(round(np.mean(accs), 3))
```
