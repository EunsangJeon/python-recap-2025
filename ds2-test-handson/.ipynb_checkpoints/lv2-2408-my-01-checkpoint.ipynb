# Q1

```python
df_1 = POWER.copy()
```

# 1-1

```python
df_1.columns
```

```python
targets = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6']
```

```python
for target in targets:
    df_1[target + '_O'] = (df_1[target] > df_1.groupby(['SEASON', 'FAC'])[target].transform(lambda x: x.quantile(0.5)) + 2 * (df_1.groupby(['SEASON', 'FAC'])[target].transform(lambda x: x.quantile(0.75)) - df_1.groupby(['SEASON', 'FAC'])[target].transform(lambda x: x.quantile(0.25)))) + 0
```

```python
df_1.head()
```

```python
df_1['OUT'] = ((df_1['F1_O'] + df_1['F2_O'] + df_1['F3_O'] + df_1['F4_O'] + df_1['F5_O'] + df_1['F6_O']) > 0)  + 0
```

```python
df_1.head()
```

# 1-2

```python
df_1.groupby(['SEASON', 'FAC'])['OUT'].mean().reset_index()['OUT'].max()
```

# Q1 Answer

```python
print(round(df_1.groupby(['SEASON', 'FAC'])['OUT'].mean().reset_index()['OUT'].max(), 4))
```

# 2-3

```python
summary = pd.DataFrame(data=['fall', 'spring', 'summer', 'winter'], columns=['season'])
```

```python
summary['mean'] = df_2.groupby('SEASON')['TOTAL'].mean().reset_index()['TOTAL']
```

```python
summary['std'] = df_2.groupby('SEASON')['TOTAL'].std().reset_index()['TOTAL']
```

```python
summary['count'] = df_2.groupby('SEASON')['TOTAL'].count().reset_index()['TOTAL']
```

```python
summary
```

```python
summary['min_z'] = summary['mean'] - (220 / (summary['count'] ** 0.5)) * 1.64
```

```python
summary['max_z'] = summary['mean'] + (220 / (summary['count'] ** 0.5)) * 1.64
```

```python
summary
```

```python
summary['conf_l'] = summary['max_z'] - summary['min_z']
```

```python
summary
```

# 2-4

```python
summary['conf_l'].max()
```

# Q2 Answer

```python
print(round(summary['conf_l'].max(), 1))
```

```python
df_3 = POWER.copy()
```

```python
df_3.shape
```

# 3-1

```python
df_3 = df_3[(df_3['F1'] != 0) & (df_3['F2'] != 0) & (df_3['F3'] != 0) & (df_3['F4'] != 0) & (df_3['F5'] != 0) & (df_3['F6'] != 0)]
```

```python
df_3.shape
```

```python
df_3 = df_3.sort_values(by=['DATE', 'FAC'])
```

```python
# df_3 = df_3.sort_values(by='FAC')
```

```python
# df_3 = df_3.reset_index()
```

```python
df_3
```

# 3-2

```python
df_3['F6_NEXT'] = df_3.groupby('FAC')['F6'].shift(-1)
```

```python
df_3.head()
```

```python
pd.notna(df_3['F6_NEXT'])
```

```python
df_3.dropna(subset=['F6_NEXT'])
```

```python
pd.notna(df_3['F6_NEXT'])
```

```python
df_3 = df_3.dropna(subset=['F6_NEXT'])
df_3.shape
```

# 3-3

```python
TRAIN3 = df_3[df_3['DATE'] < '2022-09-01'].copy()
TEST3 = df_3[df_3['DATE'] >= '2022-09-01'].copy()
```

```python
TRAIN3.shape
```

```python
TEST3.shape
```

# 3-4

```python
from sklearn.linear_model import LinearRegression
```

```python
targets = ['F1', 'F2', 'F3', 'F4', 'F5', 'F6']
```

```python
model = LinearRegression().fit(y=TRAIN3['F6_NEXT'], X=TRAIN3.loc[:, targets])
```

# 3-5

```python
pred = model.predict(X=TEST3.loc[:, targets])
```

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
mape = mean_absolute_percentage_error(y_true=TEST3['F6_NEXT'], y_pred=pred)
```

```python
mape
```

# Q3 Answer

```python
print(round(mape * 100, 3))
```
