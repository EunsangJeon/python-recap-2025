# Q2

```python
df_2 = SURVEY.copy()
```

# 2-1

```python
DF2 = df_2[df_2['AGE'] == '25-34 years old']
```

```python
DF2.shape
```

# 2-2

```python
df2_tmp = DF2.dropna(subset='SALARY')
```

```python
df2_median = df2_tmp['SALARY'].median()
```

```python
DF2['SALARY'] = DF2['SALARY'].fillna(value=df2_median)
```

```python
df2_tmp = DF2.dropna(subset='SALARY')
```

```python
df2_tmp.shape
```

```python
DF2 = DF2.sort_values(by='D_ID').reset_index(drop=True)
```

```python
DF2
```

```python
DF2['TOOL_CNT'] = DF2['TOOL'].map(lambda x: len(x.split(';')))
```

```python
DF2.head()
```

# 2-4

```python
targets = ['SATIS', 'SALARY', 'MONITORS', 'TOOL_CNT']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
for target in targets:
    DF2[target + '_S'] = StandardScaler().fit_transform(X=DF2.loc[:, [target]])
```

```python
DF2.head()
```

# 2-5

```python
targets = ['SATIS_S', 'SALARY_S', 'MONITORS_S', 'TOOL_CNT_S']
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=2, random_state=1234).fit(X=DF2.loc[:, targets])
```

```python
DF2['LABEL'] = model.predict(X=DF2.loc[:, targets])
```

```python
DF2.head()
```

```python
DF2.groupby('LABEL')['GENDER'].value_counts()
```

# Q3

```python
df_3 = SURVEY.copy()
```

# 3-1

```python
df_3 = df_3.dropna(subset=['GENDER', 'SALARY'])
```

```python
df_3.shape
```

```python
df_3['GIT_YN'] = df_3['TOOL'].map(lambda x: ('Git' in x ) + 0)
```

```python
df_3.head(15)
```

# 3-3

```python
targets = ['YEARS', 'GENDER', 'AGE']
```

```python
df_3 = pd.get_dummies(df_3, drop_first=False, dtype='int', columns=targets)
```

```python
df_3
```

# 3-4

```python
TRAIN3=df_3[df_3['D_ID'] % 3 != 0]
TEST3=df_3[df_3['D_ID'] % 3 == 0]
```

```python
TRAIN3.shape
```

```python
TEST3.shape
```

# 3-5

```python
TRAIN3.columns
```

```python
targets = ['HOBBY', 'OPENSOURCE', 'CORPORATE', 'SALARY', 'MONITORS', 'GIT_YN', 'YEARS_0-2 years', 'YEARS_12-14 years',
       'YEARS_15-17 years', 'YEARS_18-20 years', 'YEARS_21-23 years',
       'YEARS_24-26 years', 'YEARS_27-29 years', 'YEARS_3-5 years',
       'YEARS_30 or more years', 'YEARS_6-8 years', 'YEARS_9-11 years',
       'GENDER_Female', 'GENDER_Male', 'AGE_25-34 years old', 'AGE_over 35',
       'AGE_under 25']
```

```python
targets
```

```python
len(targets)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
model = DecisionTreeRegressor(max_depth=7, random_state=1234).fit(y=TRAIN3['SATIS'], X=TRAIN3.loc[:, targets])
```

# 3-6

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
pred = model.predict(X=TEST3.loc[:, targets])
```

```python
mean_absolute_percentage_error(y_true=TEST3['SATIS'], y_pred=pred) * 100
```

# Q3

```python
print(round(mean_absolute_percentage_error(y_true=TEST3['SATIS'], y_pred=pred) * 100, 1))
```
