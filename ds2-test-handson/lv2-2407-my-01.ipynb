# Q1

```python
df_1 = CULTURE.copy()
```

```python
df_1.columns
```

# 1-1

```python
targets = ['NEMP', 'NEMP_PER_HEAD', 'INT_RATIO', 'S1', 'S2', 'S3', 'S4', 'S5', 'SCORE']
```

```python
len(targets)
```

```python
cvs = []

for target in targets:
    s = df_1[target].std()
    x = df_1[target].mean()
    cv = s/x
    cvs.append(cv)
```

```python
cvs
```

```python
len(cvs)
```

# 1-2

```python
from scipy.stats import skew
```

```python
skews = []

for target in targets:
    skews.append(skew(df_1[target]))
```

```python
len(skews)
```

# 1-3

```python
from scipy.stats import pearsonr
```

```python
targets = ['NEMP', 'NEMP_PER_HEAD', 'INT_RATIO', 'S1', 'S2', 'S3', 'S4', 'S5']
```

```python
len(targets)
```

```python
pearsons = []

for target in targets:
    pearsons.append(pearsonr(x=df_1[target], y=df_1['SCORE'])[0])
```

```python
pearsons
```

```python
len(pearsons)
```

# 1-4

```python
NUM24_A = max(cvs)
NUM24_B = min(skews)
NUM24_C = max(pearsons)
NUM24_D = NUM24_A + NUM24_B + NUM24_C

NUM24_D
```

# Q1 Answer

```python
print(round(NUM24_D, 3))
```

```python
df_2 = CULTURE.copy()
```

```python
df_2.shape
```

# 2-1

```python
DF25 = df_2[df_2['F_RATIO'] != -999].copy()
```

```python
DF25.shape
```

# 2-2

```python
DF25.columns
```

```python
targets = ['NEMP', 'NEMP_PER_HEAD', 'INT_RATIO', 'F_RATIO', 'S1', 'S2', 'S3', 'S4', 'S5']
```

```python
len(targets)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
for target in targets:
    DF25[target + '_S'] = StandardScaler().fit_transform(X=DF25.loc[:, [target]])
```

```python
DF25.head()
```

# 2-3

```python
ks = [2, 3, 4]
```

```python
models = []
```

```python
tmp = []
for target in targets:
    tmp.append(target + '_S')
```

```python
targets = tmp
targets
```

```python
from sklearn.cluster import KMeans
```

```python
for k in ks:
    model = KMeans(n_clusters=k, random_state=1234).fit(X=DF25.loc[:, targets])
    models.append(model)
```

# 2-4

```python
from sklearn.metrics import silhouette_score
```

```python
sils = []

for model in models:
    pred = model.predict(X=DF25.loc[:, targets])
    sil = silhouette_score(X=DF25.loc[:, targets], labels=pred)
    sils.append(sil)
```

```python
sils
```

```python
model = models[0]
```

```python
pred = model.predict(X=DF25.loc[:, targets])
```

```python
DF25['LABEL'] = pred
```

```python
DF25.groupby('LABEL')['C_ID'].count()
```

# Q3

```python
df_3 = CULTURE.copy()
```

```python
df_3.shape
```

```python
df_3.head()
```

# 3-1

```python
tmp = df_3[df_3['F_RATIO'] != -999]
```

```python
tmp.shape
```

```python
tmp_m = tmp['F_RATIO'].mean()
tmp_m
```

```python
df_3['F_RATIO'] = df_3['F_RATIO'].map(lambda x: tmp_m if x == -999 else x)
```

```python
df_3.head()
```

# 3-2

```python
TRAIN26 = df_3[df_3['C_ID'] % 5 != 0].copy()
TEST26 = df_3[df_3['C_ID'] % 5 == 0].copy()
```

```python
TRAIN26.shape
```

```python
TEST26.shape
```

# 3-3

```python
from sklearn.linear_model import LinearRegression
```

```python
TRAIN26.columns
```

```python
targets = ['NEMP', 'NEMP_PER_HEAD', 'INT_RATIO', 'F_RATIO', 'S1', 'S2', 'S3', 'S4', 'S5']
```

```python
model = LinearRegression().fit(y=TRAIN26['SCORE'], X=TRAIN26.loc[:, targets])
```

# 3-4

```python
coef = []

for c in model.coef_:
    coef.append(abs(c))
```

```python
coef
```

```python
max(coef)
```

```python
NUM26_A = max(coef)
```

# 3-5

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
NUM26_B = 100 * mean_absolute_percentage_error(y_true=TEST26['SCORE'], y_pred=model.predict(X=TEST26.loc[:, targets]))
```

```python
NUM26_B
```

# Q3 Answer

```python
print(round(NUM26_A + NUM26_B, 4))
```
